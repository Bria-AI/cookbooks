## 2. Basic Image Generation

Generate an image using the base BRIA-4B-Adapt model.

```python
import torch
from pipeline_bria import BriaPipeline
from PIL import Image

# Load the model
pipe = BriaPipeline.from_pretrained("briaai/BRIA-4B-Adapt", 
                                   torch_dtype=torch.bfloat16,
                                   trust_remote_code=True)
pipe.to(device="cuda")  # Use CPU if no GPU is available: pipe.to(device="cpu")

# Set parameters for generation
prompt = "A beautiful landscape with mountains, a lake, and a sunset"
negative_prompt = "Logo, Watermark, Text, Ugly, Morbid, Extra fingers, Poorly drawn hands"

# Generate the image
seed = 42
generator = torch.Generator("cuda").manual_seed(seed)  # Use torch.Generator(device="cpu") for CPU
image = pipe(prompt=prompt, 
           negative_prompt=negative_prompt, 
           height=1024, 
           width=1024, 
           num_inference_steps=30,
           guidance_scale=5.0,
           generator=generator).images[0]

# Display the image
image
```

## 5. Generating Images with Fine-tuned LoRA

After fine-tuning, you can generate images using your custom LoRA weights.

```python
import torch
from pipeline_bria import BriaPipeline

# Load the model
pipe = BriaPipeline.from_pretrained("briaai/BRIA-4B-Adapt", 
                                   torch_dtype=torch.bfloat16,
                                   trust_remote_code=True)

# Load your fine-tuned LoRA weights
pipe.load_lora_weights("output_lora/", weight_name="pytorch_lora_weights.safetensors")
pipe.to(device="cuda")  # Use CPU if no GPU is available

# Set parameters for generation
prompt = "A character named CustomChar having a picnic in a meadow"
negative_prompt = "Logo, Watermark, Text, Ugly, Morbid, Extra fingers, Poorly drawn hands"

# Generate the image
seed = 42
lora_scale = 1.0  # You can adjust this to control the influence of your LoRA
generator = torch.Generator("cuda").manual_seed(seed)
image = pipe(prompt=prompt, 
           negative_prompt=negative_prompt, 
           height=1024, 
           width=1024,
           num_inference_steps=30,
           guidance_scale=5.0,
           generator=generator, 
           joint_attention_kwargs={"scale": lora_scale}).images[0]

# Display the image
image
```

## 7. Tips for Better Results

According to the BRIA-4B-Adapt documentation:

1. **Image Generation Tips:**
   - Use negative prompts to avoid unwanted elements
   - Recommended resolution: around 1M pixels (e.g., 1024x1024, 1280x768, etc.)
   - Use 30-50 inference steps for better quality
   - Use guidance_scale of 5.0

2. **Fine-tuning Tips:**
   - Use varied images while maintaining consistency in what you want the model to learn
   - Image resolution should be at least 1024x1024
   - Captions should describe unique content but use consistent descriptions or trigger words
   - Experiment with rank (default 128) - lower for simple cases, higher for finer details
   - Try different max_train_steps values and evaluate different checkpoints

## 8. Experimenting with Different Prompts and Parameters

```python
# Let's try different prompts and parameters
import torch
from pipeline_bria import BriaPipeline

# Make sure the model is loaded
if 'pipe' not in locals():
    pipe = BriaPipeline.from_pretrained("briaai/BRIA-4B-Adapt", 
                                      torch_dtype=torch.bfloat16,
                                      trust_remote_code=True)
    pipe.to(device="cuda")

# Function to generate an image with given parameters
def generate_image(prompt, negative_prompt="", height=1024, width=1024, steps=30, guidance=5.0, seed=None):
    if seed is None:
        seed = torch.randint(0, 2**32, (1,)).item()
    
    generator = torch.Generator("cuda").manual_seed(seed)
    
    print(f"Generating image with seed: {seed}")
    image = pipe(prompt=prompt, 
               negative_prompt=negative_prompt, 
               height=height, 
               width=width,
               num_inference_steps=steps,
               guidance_scale=guidance,
               generator=generator).images[0]
    
    return image, seed

# Try a different prompt
prompt = "A futuristic cityscape at night with neon lights and flying cars"
negative_prompt = "blurry, low quality, distorted, ugly"

image, seed = generate_image(prompt, negative_prompt)
image
```

```python
# Try a different aspect ratio
prompt = "A serene beach at sunset with palm trees and gentle waves"
negative_prompt = "people, blurry, low quality"

# Landscape format (16:9)
image, seed = generate_image(prompt, negative_prompt, height=768, width=1344)
image
```

```python
# Try more inference steps for higher quality
prompt = "A detailed portrait of a fantasy character with intricate armor"
negative_prompt = "blurry, low quality, distorted, ugly, extra fingers, poorly drawn hands"

# Increase steps to 50
image, seed = generate_image(prompt, negative_prompt, steps=50)
image
```
